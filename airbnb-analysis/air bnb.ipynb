```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/air_bnb_data.csv")
```

```python
df.head()
```

```python
df.columns
```

### Checking Missing Values

```python
df.isnull().sum()
```

```python
df.info()
```

### Change the datatype of Date Columns

```python
df['last review'] = pd.to_datetime(df['last review'], errors = "coerce") # Missing Value 
```

```python
df.info()
```

```python
df.fillna({'reviews per month':0, 'last review':df['last review'].min()}, inplace = True)
```

```python
df.dropna(subset = {'NAME','host name'},inplace = True)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns={"license","house_rules"}, errors = 'ignore')
```

```python
df.head()
```

```python
# Remove dollar sign and convert to float
df['price'] = df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee'] = df['price'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head(2)
```

```python
# Remove Duplicates

df.drop_duplicates(inplace = True)
```

```python
df.info()
```

```python
# Descriptive Statistics

df.describe()
```

# Visualization

### What is the distribution of listing prices?

```python
plt.figure(figsize = (10,6))
sns.histplot(df['price'],
             bins = 50,
             kde = True, # Curve
             color = 'red')
plt.title("Distrtibution of Listing Price")
plt.xlabel("Price $")
plt.ylabel("Frequency")
plt.show()
```

The histogram shows a fairly even distribution of listing prices across different price ranges, indicating no particular concentration of listings in any specific price range. The KDE line helps visualize this even spread more clearly, confirming that the dataset contains listings with a wide variety of prices.

# How are different Room Types Distribution

```python
df['room type'].unique()
```

```python
plt.figure(figsize = (8,5))
sns.countplot(x = 'room type',
              data = df,
              color = 'hotpink')
plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```

# How are Listings Distributed Across different Neighbourhood?

```python
plt.figure(figsize = (12,8))
sns.countplot(y = 'neighbourhood group',
              data = df,
              color = 'lightgreen',
              order = df['neighbourhood group'].value_counts().index
             )
plt.title('Number of Listings by Neighbourhood Group')
plt.ylabel('Neighbourhood Group')
plt.xlabel('Count')
plt.show()
```

# What is the Relationship between Price and Room Type

```python
plt.figure(figsize = (10,6))
sns.boxplot( x = 'room type',
              y = 'price',
              hue = 'room type',
              data = df,
              palette = 'Set1'
             )
plt.title('Price vs. Room Type')
plt.ylabel('Price ($)')
plt.xlabel('Room Type')
plt.legend(title = 'Room Type')
plt.show()
```

# How has the Number of Reviews Change over with Time

```python
df['last review'] = pd.to_datetime(df['last review'])
review_over_time = df.groupby(df['last review'].dt.to_period('M')).size()
review_over_time
```

```python
plt.figure(figsize = (12,6))
review_over_time.plot( kind = 'line',color = 'red')

plt.title('Number of Reviews over Time')
plt.ylabel('Number of Reviews')
plt.xlabel('Date')
plt.show()
```
